```python
import pandas as pd

# Create dataset
data = {
    "Student": ["Ananya", "Rahul", "Sravani", "Kiran",
                "Priya", "Arjun", "Divya", "Manoj"],
    "Age": [20, 21, 19, 22, 20, 21, 19, 22],
    "Marks": [85, 72, 91, 64, 78, 55, 88, 70],
    "Attendance": [92, 81, 95, 74, 88, 69, 90, 79],
    "Department": ["AI", "CSE", "AI", "ECE",
                   "CSE", "ECE", "AI", "CSE"]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)

# ---------------------------------
# Example 1: Marks greater than 80
# ---------------------------------

high_marks = df[df["Marks"] > 80]

print("Students with Marks greater than 80:")
display(high_marks)

# ---------------------------------
# Example 2: Attendance >= 90
# ---------------------------------

high_attendance = df[df["Attendance"] >= 90]

print("Students with Attendance >= 90:")
display(high_attendance)

# ---------------------------------
# Example 3: Multiple conditions
# Marks > 75 AND Attendance > 85
# ---------------------------------

good_performance = df[
    (df["Marks"] > 75) &
    (df["Attendance"] > 85)
]

print("Students with Marks > 75 AND Attendance > 85:")
display(good_performance)

# ---------------------------------
# Example 4: OR condition
# AI Department OR Marks > 85
# ---------------------------------

ai_or_high_marks = df[
    (df["Department"] == "AI") |
    (df["Marks"] > 85)
]

print("AI students OR students with Marks > 85:")
display(ai_or_high_marks)

# Number of records
print("Number of records in original dataset:", len(df))
print("Number of records after filtering:", len(good_performance))
```
